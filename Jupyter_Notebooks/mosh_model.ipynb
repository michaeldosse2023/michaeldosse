```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import joblib
from sklearn import tree
```

```python
# Importing Original Data
df = pd.read_csv('music.csv')
df.head()
```

```python
# Creating the Input set & Output set
X = df.drop(columns=['genre'])
X.head() # Input set
```

```python
# Output Data:
y = df['genre'] # Output set
y.head()
```

```python
# Preparing model using an algorithm:
model = DecisionTreeClassifier() # New instance of DecisionTreeClassifier class
model.fit(X, y) # Training the new model to learn patterns in our data
predictions = model.predict([ [21, 1], [22, 0] ]) # Predicting results
predictions
```

```python
# Measuring our model's Accuracy:
# (1) Splitting data into train & test:
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.2)

# (2) Training the model
model.fit(X_train, y_train)

# (3) Testing the model
predictions = model.predict(X_test)

# (4) Actual accuracy calculation
score = accuracy_score(y_test, predictions)
score
```

```python
# All-in-One from Gemini AI:
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
import joblib

# 1. Load your data
music_data = pd.read_csv('music.csv')
X = music_data.drop(columns=['genre'])
y = music_data['genre']

# 2. Create and Train the model
model = DecisionTreeClassifier()
model.fit(X, y)

# 3. Save the "Brain" immediately
joblib.dump(model, 'music-recommender.joblib')

print("Success! The model is trained and saved as music-recommender.joblib")
```

```python
# Reloading the model:
model = joblib.load('music-recommender.joblib')
predictions = model.predict([ [22, 0] ])
predictions
```
