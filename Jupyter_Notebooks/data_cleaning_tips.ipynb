### Re-setting Column Titles

```python
# Re-setting Column Titles:

import pandas as pd
import numpy as np

df = pd.read_csv('acti_vity.csv') # optional: ", on_bad_line='skip'"
df.head()

def wash_and_dry(input_df):
    # 1. Create a copy to protect the original data
    df_cleaned = input_df.copy() 
    
    # 2. Text Consistency: Normalize column headers
    df_cleaned.columns = [col.lower().replace(' ', '_') for col in df_cleaned.columns]
    
    # 3. CRITICAL: Return the result back to you
    return df_cleaned

    # 3. Catch the result in a new variable name (e.g., df_final)
df_final = wash_and_dry(df)

# 4. Now you can print df_final
#print(df_final.head())
df_final.to_csv(r'C:\Users\micha\OneDrive\Desktop\cleaned_acti_vity.csv') 
```

#### Trimming Data Turning to Lower Case

```python
# Removing unnecessary spaces and turning all to lower case:
# YouTube Tutorial: https://www.youtube.com/watch?v=NeJKaolLQqU

df = pd.read_csv('dataset.csv')  # optional: ", on_bad_line='skip'"
# print(df.columns.tolist())
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
print(df.columns.tolist()) # To check the cleaning result
```
