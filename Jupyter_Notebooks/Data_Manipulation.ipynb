```python
# Example of Functions that don't return
def least_difference(a, b, c):
    """Return the smallest difference between any two numbers
    among a, b and c.
    """
    diff1 = abs(a - b)
    diff2 = abs(b - c)
    diff3 = abs(a - c)
    min(diff1, diff2, diff3)
    
print(
    least_difference(1, 10, 100),
    least_difference(1, 10, 10),
    least_difference(5, 6, 7),
)
```

```python
# Functions Applied to Functions; here 'mult_by_five' is applied to 'call'
def mult_by_five(x):
    return 5 * x

def call(fn, arg):
    """Call fn on arg"""
    return fn(arg)

def squared_call(fn, arg):
    """Call fn on the result of calling fn on arg"""
    return fn(fn(arg))

print(
    call(mult_by_five, 1),
    squared_call(mult_by_five, 1), 
    sep='\n', # '\n' is the newline character - it starts a new line
)
```

```python
# Comparison Operations:
def can_run_for_president(age):
    """Can someone of the given age run for president in the US?"""
    # The US Constitution says you must be at least 35 years old
    return age >= 35

print("Can a 19-year-old run for president?", can_run_for_president(19))
print("Can a 45-year-old run for president?", can_run_for_president(45))
```

### AI Amended Codes for NG LMS Schema Items

```python
# 1. Organisation Table (Create this first so others can reference it)
CREATE TABLE organisation (
    orgid VARCHAR(20) PRIMARY KEY,
    name VARCHAR(100),
    email VARCHAR(100) UNIQUE,
    telephone VARCHAR(20) # Use VARCHAR for phones to handle leading zeros
);

# 2. Users Table
CREATE TABLE users (
    employee_number INT PRIMARY KEY,
    ni_number VARCHAR(9),      
    title VARCHAR(50),              
    first_name VARCHAR(250),
    preferred_name VARCHAR(250),    
    last_name VARCHAR(250),
    dob DATE,
    email_address VARCHAR(250) UNIQUE,
    position_title VARCHAR(250),
    user_role VARCHAR(100), # Renamed from 'Role' to avoid reserved word
    staff_group VARCHAR(250),              
    division VARCHAR(250),
    org_unit_group VARCHAR(250), # Renamed from 'Group'
    directorate VARCHAR(250),
    department_id VARCHAR(20) REFERENCES organisation(orgid), # FOREIGN KEY
    location VARCHAR(250)
);      

# 3. Course Table 
CREATE TABLE course (
    courseid VARCHAR(20) PRIMARY KEY,
    title VARCHAR(250),
    status VARCHAR(50),
    visibility VARCHAR(50),
    length_minutes INT,
    start_date DATE
);

# 4. Classes Table
CREATE TABLE class (
    classid VARCHAR(20) PRIMARY KEY,
    courseid VARCHAR(20) REFERENCES course(courseid), # FOREIGN KEY: Links class to a course
    class_type VARCHAR(50),
    start_date DATE,
    capacity INT
);

# 5. Enrolment Table
CREATE TABLE enrolment (
    enrolid VARCHAR(20) PRIMARY KEY,
    employee_number INT REFERENCES users(employee_number), # FOREIGN KEY: Links to User
    courseid VARCHAR(20) REFERENCES course(courseid),     # FOREIGN KEY: Links to Course
    enrolment_date DATE,
    status VARCHAR(50)
);
```
