### Building a Cipher

```python
custom_key = "happycoding" #"python"
text = "mrttaqrhknsw thpggrur" 

def vigenere(message, key, direction=1):
    key_index = 0
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    final_message = '' # was encrypted_text

    for char in message.lower(): 
        # (1) Appende any non-letter character to the message
        if not char.isalpha(): # was "if char == ' ':"
            final_message += char
        else:
            
            # (2) Find the right ley character to encode/decode:
            key_char = key[key_index % len(key)]
            key_index += 1
            # (3) Define the offset and the encrypted/decrypted letter:
            offset = alphabet.index(key_char)
            index = alphabet.find(char)
            new_index = (index + offset * direction) % len(alphabet)
            final_message += alphabet[new_index]
    return final_message
#    print("plain text:", message)
#    print('encrypted_text:', encrypted_text)

#Key Takeaways (ADVICE by AI):
#return should almost always be at the very end of your function (aligned with the for loop, not inside it).
#Calling the function must happen at the "zero" indentation level (the far left) to actually trigger the code.

# Function called & stored in a new variable (NO LONGER REQUIRED):
#encryption = encrypt(text, custom_key) # initial argument "1" removed
#print(encryption)

#decryption = decrypt(text, custom_key)
#print(f' \n Decrypted text: {decryption}\n')

def encrypt(message, key):
    return vigenere(message, key)

def decrypt(message, key):
    return vigenere(message, key, -1)
    
print(f'\n Encrypted text: {text}')
print(f'Key: {custom_key}')
decryption = decrypt(text, custom_key)
print(f' \n Decrypted text: {decryption}\n')
```

### Calculator CODES Explained

```python
# (1) Define the main function:
def start_calculator():
    print("--- Simple Calculator ---")
    try:
        num1 = float(input("Enter first number: "))
        op = input("Enter operator (+, -, *, /): ")
        num2 = float(input("Enter second number: "))

        if op == '+': # ADD
            print(f"Result: {num1 + num2}")
        elif op == '-': # SUBTRACT
            print(f"Result: {num1 - num2}")
        elif op == '*': # MULTIPLY
            print(f"Result: {num1 * num2}")
        elif op == '/': # DIVIDE
            print(f"Result: {num1 / num2}")
        else:
            print("Invalid operator!")

    except ValueError:
        print("Error: Please enter numbers only.")

    # This line prevents the .exe from closing instantly!
    input("\nPress Enter to exit...")


if __name__ == "__main__":
    start_calculator()
```

#### VAT Calculation in Python

```python
# Calculating VAT:

# Defining the function
def calculate_vat(amount, vat_rate=0.20): # wiht 2 parameters: amount & VAT rate
    vat = amount * vat_rate # VAT formula
    return vat

amounts = [100, 250, 1500] # Sample amounts to calculate as a list

for amt in amounts: # The FOR loop to go through the sample amounts and return the VAT values
    vat_amount = calculate_vat(amt) # Passing the function call in a new variable 
    print(f"Original amount: £{amt:.2f} -> VAT: £{vat_amount:.2f}") # Printing the results by diplaying the initial amount ...
                                                                    # ... and the VAT values with 2 decimal places (.2f)
    
```
